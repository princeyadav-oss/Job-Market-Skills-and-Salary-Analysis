```python
# 1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from io import BytesIO
import base64, gzip

sns.set_theme(style="whitegrid")
print("Libraries imported successfully.")
```

```python
# 2. Load the embedded Kaggle dataset
# No file upload is required.
encoded_data = ''
csv_data = gzip.decompress(base64.b64decode(encoded_data))
df = pd.read_csv(BytesIO(csv_data))

print("Dataset loaded successfully!")
print("Shape:", df.shape)
df.head()
```

```python
# 3. Dataset Overview
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum())
```

```python
# 4. Data Cleaning
df_clean = df.copy()

# Remove duplicate records
before = len(df_clean)
df_clean = df_clean.drop_duplicates()
print("Duplicates removed:", before - len(df_clean))

# Fill numeric missing values with median
numeric_cols = df_clean.select_dtypes(include=np.number).columns
for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Fill text missing values
text_cols = df_clean.select_dtypes(include="object").columns
for col in text_cols:
    df_clean[col] = df_clean[col].fillna("Unknown")

print("Cleaned shape:", df_clean.shape)
display(df_clean.head())
```

```python
import plotly.graph_objects as go
import numpy as np

# Salary data
salary = df_clean["salary_usd"].dropna()

# 30 bins
counts, bins = np.histogram(salary, bins=30)

# Bin centers
centers = (bins[:-1] + bins[1:]) / 2
width = bins[1] - bins[0]

# Different colors
colors = [
    "#FF6B6B", "#4ECDC4", "#45B7D1", "#96CEB4",
    "#FFEAA7", "#DDA0DD", "#FF9F43", "#54A0FF",
    "#5F27CD", "#00D2D3", "#FF6348", "#2ED573",
    "#FFA502", "#1E90FF", "#A55EEA"
]

# Repeat colors until 30 bars
bar_colors = [
    colors[i % len(colors)]
    for i in range(30)
]

# Animation frames
frames = []

for i in range(1, 31):

    frames.append(
        go.Frame(
            data=[
                go.Bar(
                    x=centers[:i],
                    y=counts[:i],
                    width=width * 0.85,

                    # Different color for every bar
                    marker=dict(
                        color=bar_colors[:i],
                        line=dict(
                            color="white",
                            width=1
                        )
                    )
                )
            ],
            name=str(i)
        )
    )

# First frame
fig = go.Figure(
    data=[
        go.Bar(
            x=centers[:1],
            y=counts[:1],
            width=width * 0.85,
            marker=dict(
                color=[bar_colors[0]],
                line=dict(
                    color="white",
                    width=1
                )
            )
        )
    ],
    frames=frames
)

# Layout
fig.update_layout(

    title="💰 Animated Salary Distribution",

    xaxis_title="Salary (USD)",
    yaxis_title="Number of Jobs",

    xaxis=dict(
        range=[bins[0], bins[-1]
        ]
    ),

    yaxis=dict(
        range=[0, counts.max() * 1.2]
    ),

    template="plotly_white",

    # PLAY BUTTON
    updatemenus=[
        {
            "type": "buttons",
            "direction": "left",

            "x": 0.1,
            "y": 1.15,

            "showactive": False,

            "buttons": [
                {
                    "label": "▶ Play",
                    "method": "animate",

                    "args": [
                        None,
                        {
                            "frame": {
                                "duration": 300,
                                "redraw": True
                            },

                            "transition": {
                                "duration": 150
                            },

                            "fromcurrent": False
                        }
                    ]
                }
            ]
        }
    ]
)

fig.show()
```

```python
# 6. Job Role Analysis - Colorful Auto Animation

import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import HTML, display

# Top 10 Job Roles
role_col = "job_title"
top_roles = df_clean[role_col].value_counts().head(10)

display(top_roles)

# Sort ascending for horizontal chart
roles = top_roles.sort_values(ascending=True)

# Different colors
colors = [
    "#FF4B5C",
    "#00B8D9",
    "#7B61FF",
    "#00C853",
    "#FFAB00",
    "#E84393",
    "#0984E3",
    "#6C5CE7",
    "#00CEC9",
    "#FD79A8"
]

# Create frames
frames = []

for i in range(1, len(roles) + 1):

    # First i values visible, remaining values = 0
    animated_values = [
        roles.iloc[j] if j < i else 0
        for j in range(len(roles))
    ]

    frames.append(
        go.Frame(
            data=[
                go.Bar(
                    x=animated_values,
                    y=roles.index,
                    orientation="h",

                    marker=dict(
                        color=colors,
                        line=dict(
                            color="white",
                            width=2
                        )
                    ),

                    text=[
                        str(roles.iloc[j]) if j < i else ""
                        for j in range(len(roles))
                    ],

                    textposition="outside"
                )
            ],
            name=f"frame{i}"
        )
    )

# Initial chart
fig = go.Figure(
    data=[
        go.Bar(
            x=[roles.iloc[0]] + [0] * (len(roles) - 1),
            y=roles.index,
            orientation="h",

            marker=dict(
                color=colors,
                line=dict(
                    color="white",
                    width=2
                )
            ),

            text=[str(roles.iloc[0])] + [""] * (len(roles) - 1),
            textposition="outside"
        )
    ],
    frames=frames
)

# Layout
fig.update_layout(
    title="🌈 Top 10 Job Roles by Number of Postings",

    xaxis=dict(
        title="Number of Postings",
        range=[0, roles.max() * 1.15]
    ),

    yaxis=dict(
        title="Job Role",
        categoryorder="array",
        categoryarray=roles.index
    ),

    height=650,

    template="plotly_white",

    font=dict(size=13)
)

# Frame names
frame_names = [f"frame{i}" for i in range(1, len(roles) + 1)]

# Auto Play
html = pio.to_html(
    fig,
    full_html=False,
    include_plotlyjs=True,

    post_script=f"""
    setTimeout(function() {{
        Plotly.animate(
            '{'{plot_id}'}',
            {frame_names},
            {{
                frame: {{
                    duration: 800,
                    redraw: true
                }},
                transition: {{
                    duration: 400
                }},
                mode: 'immediate',
                fromcurrent: false
            }}
        );
    }}, 1000);
    """
)

display(HTML(html))
```

```python
# 7. Experience Level vs Salary

import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import HTML, display

experience_col = "experience_level"
salary_col = "salary_usd"

# Salary statistics
exp_salary = (
    df_clean.groupby(experience_col)[salary_col]
    .agg(["count", "mean", "median"])
    .sort_values("mean", ascending=False)
)

display(exp_salary)

# Experience levels
levels = exp_salary.index.tolist()
mean_salary = exp_salary["mean"].tolist()

# Colors
colors = [
    "#FF4B5C",
    "#00B8D9",
    "#7B61FF",
    "#00C853",
    "#FFAB00"
]

# Create animation frames
frames = []

for i in range(1, len(levels) + 1):

    values = [
        mean_salary[j] if j < i else 0
        for j in range(len(levels))
    ]

    text_values = [
        f"${mean_salary[j]:,.0f}" if j < i else ""
        for j in range(len(levels))
    ]

    frames.append(
        go.Frame(
            data=[
                go.Bar(
                    x=levels,
                    y=values,
                    marker=dict(
                        color=colors[:len(levels)],
                        line=dict(
                            color="white",
                            width=2
                        )
                    ),
                    text=text_values,
                    textposition="outside"
                )
            ],
            name=f"frame{i}"
        )
    )

# Initial chart
fig = go.Figure(
    data=[
        go.Bar(
            x=levels,
            y=[mean_salary[0]] + [0] * (len(levels) - 1),

            marker=dict(
                color=colors[:len(levels)],
                line=dict(
                    color="white",
                    width=2
                )
            ),

            text=[
                f"${mean_salary[0]:,.0f}"
            ] + [""] * (len(levels) - 1),

            textposition="outside"
        )
    ],

    frames=frames
)

# Layout
fig.update_layout(
    title="💰 Average Salary by Experience Level",

    xaxis=dict(
        title="Experience Level"
    ),

    yaxis=dict(
        title="Average Salary (USD)",
        range=[0, max(mean_salary) * 1.20]
    ),

    height=550,

    template="plotly_white",

    font=dict(size=13),

    # Remove unnecessary toolbar buttons
    modebar=dict(
        remove=[
            "zoom",
            "pan",
            "select",
            "lasso2d",
            "autoScale",
            "resetScale"
        ]
    )
)

# Frame names
frame_names = [
    f"frame{i}"
    for i in range(1, len(levels) + 1)
]

# AUTO PLAY
html = pio.to_html(
    fig,
    full_html=False,
    include_plotlyjs=True,

    post_script=f"""
    setTimeout(function() {{
        Plotly.animate(
            '{{plot_id}}',
            {frame_names},
            {{
                frame: {{
                    duration: 900,
                    redraw: true
                }},
                transition: {{                    duration: 400
                }},
                mode: 'immediate',
                fromcurrent: false
            }}
        );
    }}, 1000);
    """
)

display(HTML(html))
```

```python
# Salary vs Years of Experience - Animated Scatter Plot

import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import HTML, display

# Remove missing values
scatter_df = df_clean[
    ['years_experience', 'salary_usd']
].dropna().sort_values('years_experience')

x = scatter_df['years_experience'].tolist()
y = scatter_df['salary_usd'].tolist()

# Create frames
frames = []

step = max(1, len(x) // 30)

for i in range(step, len(x) + 1, step):
    frames.append(
        go.Frame(
            data=[
                go.Scatter(
                    x=x[:i],
                    y=y[:i],
                    mode='markers',
                    marker=dict(
                        size=8,
                        color=y[:i],
                        colorscale='Turbo',
                        showscale=True,
                        colorbar=dict(title="Salary")
                    ),
                    text=[
                        f"Experience: {x[j]} years<br>"
                        f"Salary: ${y[j]:,.0f}"
                        for j in range(i)
                    ],
                    hoverinfo='text'
                )
            ],
            name=f"frame{i}"
        )
    )

# Final frame if needed
if len(x) % step != 0:
    frames.append(
        go.Frame(
            data=[
                go.Scatter(
                    x=x,
                    y=y,
                    mode='markers',
                    marker=dict(
                        size=8,
                        color=y,
                        colorscale='Turbo',
                        showscale=True,
                        colorbar=dict(title="Salary")
                    )
                )
            ],
            name="final"
        )
    )

# Initial chart
fig = go.Figure(
    data=[
        go.Scatter(
            x=x[:step],
            y=y[:step],
            mode='markers',
            marker=dict(
                size=8,
                color=y[:step],
                colorscale='Turbo',
                showscale=True,
                colorbar=dict(title="Salary")
            )
        )
    ],
    frames=frames
)

fig.update_layout(
    title="💰 Salary vs Years of Experience",
    xaxis=dict(
        title="Years of Experience"
    ),
    yaxis=dict(
        title="Salary (USD)"
    ),
    height=600,
    template="plotly_white",
    font=dict(size=13)
)

# Auto-start animation
frame_names = [frame.name for frame in frames]

html = pio.to_html(
    fig,
    full_html=False,
    include_plotlyjs=True,
    post_script=f"""
    setTimeout(function() {{
        Plotly.animate(
            '{'{plot_id}'}',
            {frame_names},
            {{
                frame: {{
                    duration: 120,
                    redraw: false
                }},
                transition: {{
                    duration: 80
                }},
                mode: 'immediate',
                fromcurrent: false
            }}
        );
    }}, 1000);
    """
)

display(HTML(html))
```
